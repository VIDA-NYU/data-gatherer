```python
import os
# MUST set this BEFORE importing torch/transformers
os.environ['PYTORCH_MPS_HIGH_WATERMARK_RATIO'] = '0.0'

import sklearn
import numpy as np
import pandas as pd
import time, re, json
from sklearn.model_selection import train_test_split
```

```python
ground_truth_complete = "scripts/Local_model_finetuning/ground_truth/gt_dataset_info_extraction_from_snippet.xlsx"
ground_truth_no_dspage = "scripts/Local_model_finetuning/ground_truth/gt_dataset_info_no_dspage_extraction_from_snippet.xlsx"
```

```python
df_1 = pd.read_excel(ground_truth_no_dspage)
df_2 = pd.read_excel(ground_truth_complete)
len(df_1), len(df_2)
```

```python
lst, exact, cnt = [], [], 0
for i in range(len(df_1)):
    www, ext = False, False
    dspage_ref_input = df_2.iloc[i]['input_text']
    raw = df_2.iloc[i]['output_text']
    if pd.isna(raw):
        print(f"Row {i} has missing output_text: {dspage_ref_input}")
        continue
    dspage_text = json.loads(raw)
    dspage_output = [item.get('dataset_webpage') for item in dspage_text.get('datasets', []) if item.get('dataset_webpage')!= 'n/a']    
    cnt += len(dspage_output)

    if re.search(r'www', dspage_ref_input):
        lst.append((i, dspage_ref_input))
        www = True
    
    for ds_url in dspage_output:
        if ds_url.lower() in dspage_ref_input.lower():
            exact.append((i, dspage_ref_input))
            ext = True
            break

print(f"Total www found: {len(lst)}")
print(f"Total exact urls found: {len(exact)}")
print(f"Total non-null dataset urls extracted: {cnt}")
```
